```python
import torch
import numpy as np
import nibabel as nb

subj = '8'
whole_region = torch.load(f'/mnt/sdc/project/Brain_Network/fMRI2img/Dataset/preprocessed_data/subject{subj}/nsd_imagery_whole_brain.pt')


original_shape = nb.load(f'/mnt/sdc/project/Brain_Network/fMRI2img/Dataset/nsddata/ppdata/subj0{subj}/func1pt8mm/roi/nsdgeneral.nii.gz').get_fdata().shape 
inflated_mask = nb.load(f'/mnt/sdc/project/Brain_Network/fMRI2img/Dataset/nsddata/ppdata/subj0{subj}/func1pt8mm/roi/brainmask_inflated_1.0.nii.gz').get_fdata()
loc = np.where(inflated_mask!=0)

reconstructed = torch.zeros((original_shape[0], original_shape[1], original_shape[2], whole_region.shape[0]), dtype=torch.float32)


mask_indices = torch.from_numpy(inflated_mask.flatten()).nonzero().squeeze()

for i in range(whole_region.shape[0]): 
    temp_slice = torch.zeros_like(reconstructed[:,:,:,0], dtype=torch.float32) 
    temp_slice.put_(mask_indices, whole_region[i])  
    reconstructed[..., i] = temp_slice.view(original_shape)  

print(reconstructed.shape)
print(np.where(reconstructed.max()))
torch.save(reconstructed, f'dataset_3d/subj{subj}_reconstructed_data.pt')
print('LOCATION OF MAX in 3d:',np.where(reconstructed==reconstructed.max()))
print('MAX VALUE of reconstructed:',reconstructed.max())
print('LOCATION OF MAX in 1d:',np.where(whole_region==whole_region.max()))
print('MAX VALUE of 1d value:',whole_region.max())
print('whole_region.mean:',whole_region.mean())
print('reconstructed.mean:',reconstructed.mean())
```
